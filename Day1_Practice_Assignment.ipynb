```python
'''Q8. Temperature Converter
Take a temperature in Celsius from the user and print it in Fahrenheit: F = (C * 9/5) + 32.
Then take a temperature in Fahrenheit and print it in Celsius.'''

#INPUT: Temperature in Celsius and Fahrenheit
#PROCESS: Convert Temperature in Celsius to Fahrenheit anf Fahrenheir to Celsius
#OUTPUT: Temperature in Fahrenheit and Celsius

#Celsius to Fahrenheit
temp=int(input("Enter temperature in Celsius:"))
fah=(temp*9/5)+32
print(f"Temperature in Fahrenheit: {fah}")

#Fahrenheit to Celsius
temp=int(input("Enter temperature in Fahrenheit:"))
cel=(temp-32)*5/9
print(f"Temperature in Celsius: {cel}")
```

```python
'''Q10. Circle Calculations
Take the radius of a circle from the user.
Print its area (pi * r * r) and circumference (2 * pi * r). Use pi = 3.14159.'''

#INPUT: radius of circle
#PROCESS: calculate area and circumference of circle
#OUTPUT: area and circumference

radius=int(input("Enter Radius: "))
pi=3.14159
area=pi*radius*radius
circumference=2*pi*radius
print(f"Area: {area:.2f}")
print(f"Circumference: {circumference:.2f}")
```

```python
'''Q11. Seconds to H:M:S
Take a total number of seconds from the user.
Using // and %, convert and print it in the format hours, minutes, seconds.'''

#INPUT: total number of seconds
#PROCESS: use // and % to calculate hours, minutes, and seconds
#OUTPUT: print the time in hours, minutes, and seconds

total_seconds = int(input("Enter total seconds: "))
hours = total_seconds // 3600
minutes = (total_seconds % 3600) // 60
seconds = total_seconds % 60
print(f"{hours} h {minutes} m {seconds} s")
```

```python
'''Q12. Sum of Digits of a Two-Digit Number
Take a two-digit number from the user.
Extract the tens digit (n // 10) and the units digit (n % 10) and print their sum.'''

#INPUT: a two-digit number
#PROCESS: extract the tens digit using // and the units digit using %, then add them
#OUTPUT: print the sum of the two digits

n = int(input("Enter a two-digit number: "))
tens = n // 10
units = n % 10
sum_of_digits = tens + units
print(f"Sum of digits = {sum_of_digits}")
```

```python
'''Q13. Minutes to Hours and Minutes
Take a number of minutes from the user.
Print how many whole hours and remaining minutes that is, using // and %.'''

#INPUT: total number of minutes
#PROCESS: use // to calculate whole hours and % to calculate remaining minutes
#OUTPUT: print the hours and remaining minutes

minutes = int(input("Enter total minutes: "))
hours = minutes // 60
remaining_minutes = minutes % 60
print(f"{hours} hours {remaining_minutes} minutes")
```

```python
'''Q15. String Basics
Take a single word from the user.
Print its length, the word in UPPERCASE, in lowercase, and its first and last character.'''

#INPUT: a single word from the user
#PROCESS: find its length, convert it to uppercase and lowercase, and access the first and last character
#OUTPUT: print the length, uppercase, lowercase, first character, and last character

word=input("Enter word: ")
print("Length:",len(word))
print("Uppercase:",word.upper())
print("Lowercase:",word.lower())
print("First Character:",word[0])
print("Last Character:",word[-1])
```

```python
'''Q16. String Slicing
Take a word from the user.
Print the word reversed (using slicing s[::-1]) and print every second character (s[::2]).'''

#INPUT: a word from the user
#PROCESS: use [::-1] to reverse the word and [::2] to get every second character
#OUTPUT: print the reversed word and every second character

word=input("Enter word: ")
reverse=word[::-1]
second=word[::2]
print("Reverse:",reverse)
print("Every Second Character:",second)
```
