```python
'''Q1. Write and Read a File
Take three lines of text from the user and write them to a file notes.txt.
Then open the file and print its full contents.'''

with open("notes.txt", "w") as file:
    for i in range(3):
        line = input("Enter a line of text: ")
        file.write(line + "\n")
with open("notes.txt", "r") as file:
    print(file.read())
```

```python
'''Q4. Average Marks from CSV (Plain Python)
Read students.csv without Pandas.
Using an accumulator, compute and print the average of the marks column.'''

file = open("students.csv", "w")

file.write("Name,Age,Marks\n")
file.write("Rahul,20,85\n")
file.write("Priya,21,90\n")
file.write("Amit,19,78\n")

file.close()

total = 0
count = 0

with open("students.csv", "r") as f:
    header = f.readline()

    for line in f:
        parts = line.strip().split(",")
        marks = int(parts[2])
        total += marks
        count += 1

print("Average marks:", total / count)
```

```python
'''Q5. Find the Topper (Plain Python)
Read students.csv without Pandas and print the name of the student with the highest marks.'''

highest_marks = 0
topper = ""

with open("students.csv", "r") as f:
    next(f)

    for line in f:
        parts = line.strip().split(",")
        name = parts[0]
        marks = int(parts[2])

        if marks > highest_marks:
            highest_marks = marks
            topper = name

print("Topper:", topper)
```

```python
'''Q16. Add a Pass/Fail Column
Add a new column Add a new column 'result' that is 'Pass' if marks >= 50 else 'Fail'.
Print the updated DataFrame (hint: define a function and use .apply()).'''

def check_result(marks):
    if marks >= 50:
        return "Pass"
    else:
        return "Fail"

df["result"] = df["Marks"].apply(check_result)

print(df)
```

```python
'''Q19. First Machine Learning Model
Using scikit-learn LinearRegression, train a model with hours_studied as the feature and marks as the target.
Predict the marks for a student who studies 6 hours and for one who studies 10 hours.
Print the learned relationship (intercept and coefficient).'''

from google.colab import files

uploaded = files.upload()

import pandas as pd
from sklearn.linear_model import LinearRegression

df = pd.read_csv("Student.csv")


X = df[["time_study"]]
y = df["Marks"]

model = LinearRegression()
model.fit(X, y)

predictions = model.predict([[6], [10]])

print("Predicted marks for 6 hours:", predictions[0])
print("Predicted marks for 10 hours:", predictions[1])

print("Intercept:", model.intercept_)
print("Coefficient:", model.coef_[0])
```
