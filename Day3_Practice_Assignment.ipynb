```python
'''Q6. Common and Unique Elements
Given two lists, use sets to print the elements common to both, and the elements only in the first.'''

list1=[1,2,6,3,7,9]
list2=[4,5,8,2,9,1]

print("Common elements: ",set(list1).intersection(list2))
print("Unique elements: ",set(list1).difference(list2))

print("=========================================")

set1=set(list1)
set2=set(list2)

print("Common elements: ",set1&set2)
print("Unique elements: ",set1-set2)
```

```python
'''Q9. Student Marks Dictionary
Create a dictionary of at least 4 students mapping name -> marks.
Print each student name-wise, then find and print the topper (highest marks).'''

students={
    "Ram":88,
    "Sham":89,
    "Anil":98,
    "Sunil":92,
}

print(students)

print("===============")

for name,marks in students.items():
  print(name,":",marks)

print("===============")
topper=max(students,key=students.get)
print("Topper: ",topper)
print("Highest marks:", students[topper])
```

```python
'''Q14. Function: factorial
Write a function factorial(n) that returns n! using a loop.
Call it for 5 and 7.'''

def factorial(n):
  if n==0 or n==1:
    return 1
  else:
    return n*factorial(n-1)

print("Factorial of 5 is:",factorial(5))
print("Factorial of 7 is:",factorial(7))
```

```python
'''Q15. Function: is_prime and Primes in a Range
Write a function is_prime(n) that returns True/False.
Use it in a loop to print all prime numbers between 1 and 50.'''

def is_prime(n):
  if n<2:
    return False
  for i in range(2,n):
    if n%i==0:
      return False
  return True


for num in range(1,51):
  if is_prime(num):
    print(num)
```

```python
'''Q16. Function: is_palindrome
Write a function is_palindrome(word) that returns True if the word reads the same backwards.
Test it on level, python, and madam.'''

def is_palindrome(word):
  if word==word[::-1]:
    return True
  else:
    return False

print("Is level a palindrome? ",is_palindrome("level"))
print("Is python a palindrome? ",is_palindrome("python"))
print("Is madam a palindrome? ",is_palindrome("madam"))
```

```python
'''Q18. Function Returning Multiple Values
Write a function stats(nums) that returns the minimum, maximum, and average of a list.
Call it and print all three values.'''

def stats(num__s):
  minimum=min(num__s)
  maximum=max(num__s)
  average=sum(num__s)/len(num__s)
  return minimum, maximum, average

num__s=[10, 20, 30, 40, 50]
minimum,maximum,average=stats(num__s)
print("Minimum: ",minimum)
print("Maximum: ",maximum)
print("Average: ",average)
```

```python
'''Q19. Function: count_vowels
Write a function count_vowels(text) that returns how many vowels are in the text.
Test it on a sentence taken from the user.'''

def count_vowels(text):
  count=0
  for ch in text:
    if ch in "aeiouAEIOU":
      count+=1

  return count

text=input("Enter a sentence: ")

print("No. of vowels in given sentence: ",count_vowels(text))
```

```python
'''Q20. Grade Function Applied to Many Students
Write a function grade(marks) that returns 'A' / 'B' / 'C' / Fail' (reuse Day 2 logic).
Given a dictionary of student -> marks, loop through it and print each student's grade.'''

def grade(marks):
  if marks>=90:
    return 'A'
  elif marks>=80:
    return 'B'
  elif marks>=70:
    return 'C'
  else:
    return 'Fail'

stud={
    "Rahul": 85,
    "Priya": 92,
    "Amit": 45,
    "Sneha": 78
  }

for name, marks in stud.items():
    print(name,":",grade(marks))
```
