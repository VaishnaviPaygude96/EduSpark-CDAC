```python
'''Q5. Leap Year Check
Take a year and decide if it is a leap year.
Rule: divisible by 4 AND (not divisible by 100 OR divisible by 400).'''

#INPUT: Take a year from the user.
#PROCESS: A year is a leap year if it is divisible by 4 and either not divisible by 100 or divisible by 400.
#OUTPUT: Print whether the year is a leap year or not.

year=int(input("Enter a year:"))
if year%4==0:
  if year%100==0:
    if year%400==0:
      print("f{year} if leap year")
    else:
      print(f"{year} is not leap year")
  else:
    print(f"{year} is leap year")
else:
  print(f"{year} is not leap year")
```

```python
'''Q12. Factorial
Take a number N and print N! (1 * 2 * ... * N) using a loop.'''

#INPUT: Take a number N.
#PROCESS: Multiply all numbers from 1 to N using a loop.
#OUTPUT: Print N!.

n=int(input("Enter a number:"))
fact=1
for i in range(1,n+1):
  fact*=i
print(f"Factorial of {n} is {fact}")
```

```python
'''Q13. Count the Digits
Take a number and count how many digits it has using a while loop (hint: n = n // 10).'''

#INPUT: Take a number from the user.
#PROCESS: Repeatedly divide the number by 10 using // and count how many times this happens.
#OUTPUT: Print the number of digits.

n=int(input("Enter a number: "))
count=0
num=n
while num>0:
  num=num//10
  count+=1

print(f"Number of digits in {n} is {count}")
```

```python
'''Q14. Reverse a Number
Take a number and print it reversed using a while loop (hint: last digit = n % 10).'''

#INPUT: Take a number from the user.
#PROCESS: Get the last digit using % 10, add it to the reversed number, and remove the last digit using // 10.
#OUTPUT: Print the reversed number.

num=int(input("Enter a number: "))
rev=0
while num>0:
  last_digit=num%10
  rev=rev*10+last_digit
  num=num//10

print(f"Reverse of {n} is {rev}")
```

```python
'''Q15. Prime Number Check
Take a number and print whether it is PRIME.
Loop from 2 to n-1; if any value divides it evenly, it is not prime (use break).'''

#INPUT: Take a number from the user.
#PROCESS: Check whether any number from 2 to n-1 divides the number. Use break if a divisor is found.
#OUTPUT: Print Prime or Not Prime.

num=int(input("Enter a number: "))
for i in range(2,num):
  if num%i==0:
    print(f"{num} is not prime number")
    break
else:
  print(f"{num} is prime number")
```

```python
'''Q16. Fibonacci Series
Take a count and print that many terms of the Fibonacci series: 0 1 1 2 3 5 8 ...'''

#INPUT: Take the number of terms to print.
#PROCESS: Start with 0 and 1. Repeatedly calculate the next term by adding the previous two terms.
#OUTPUT: Print the required number of Fibonacci terms.

n=int(input("Enter a number upto want fibinacci series: "))
a=0
b=1

for i in range(n):
  print(a, end=" ")
  a,b=b,a+b
```

```python
'''Q17. Sum of Even Numbers in a Range
Take two numbers (start and end) and print the sum of only the even numbers between them.'''

#INPUT: Take the starting and ending numbers.
#PROCESS: Loop through the range and add only the numbers divisible by 2.
#OUTPUT: Print the sum of even numbers.

num1=int(input("Enter start number: "))
num2=int(input("Enter end number: "))
sum=0
for i in range(num1,num2+1):
  if i%2==0:
    sum=sum+i
print(sum)

```

```python
'''Q20. Sum Until Zero (Sentinel Loop)
Keep asking the user for numbers using a while loop.
Stop when they enter 0 (use break).
Then print the total sum and the count of numbers entered.'''

#INPUT: Keep taking numbers from the user until the user enters 0.
#PROCESS: Use a while loop. Add each number to total and increase count. Use break when the input is 0.
#OUTPUT: Print the total sum and count of numbers entered.

sum=0
count=0
while True:
  n=int(input("Enter a number: "))
  if n==0:
    break
  sum=sum+n
  count=count+1
print(f"Sum is {sum}")
print(f"Count is {count}")
```
